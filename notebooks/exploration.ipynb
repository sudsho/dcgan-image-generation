# DCGAN exploration

Quick walkthrough: load the trained generator, sample some faces, do a latent walk between two random z.

```python
import os, sys
sys.path.append('..')
import torch
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from src.model import Generator
from src.sample import random_samples, interpolate
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
G = Generator(latent_dim=100, ngf=64, channels=3).to(device)
ckpt = torch.load('../artifacts/checkpoints/latest.pt', map_location=device)
G.load_state_dict(ckpt['G'])
G.eval();
```

## Random samples

```python
imgs = random_samples(G, 64, 100, device)
grid = make_grid((imgs + 1) / 2, nrow=8)
plt.figure(figsize=(8, 8))
plt.imshow(grid.permute(1, 2, 0))
plt.axis('off');
```

## Latent walk between two random z

```python
imgs = interpolate(G, 10, 100, device)
grid = make_grid((imgs + 1) / 2, nrow=10)
plt.figure(figsize=(20, 2))
plt.imshow(grid.permute(1, 2, 0))
plt.axis('off');
```
